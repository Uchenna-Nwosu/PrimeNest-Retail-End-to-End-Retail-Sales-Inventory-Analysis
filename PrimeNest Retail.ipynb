# - Sales by Product Category

```python
#This moves the analysis from individual products to the bigger picture(quantity): which categories are generating the most sales volume?
```

```python
"""
Because order_details has the quantity and products has the category_id, a merge of tables would be required. 

Then we connect the category information.
```

```python
category_sales = (
    order_details
    .merge(
        products[['product_id', 'category_id']],
        on='product_id',
        how='left'
    )
    .merge(
        categories[['category_id', 'category_name']],
        on='category_id',
        how='left'
    )
    .groupby('category_name')['quantity']
    .sum()
    .reset_index()
    .sort_values('quantity', ascending=False)
)

category_sales
```

```python
"""
- Sales by Product Category: Observation
Electronics is the strongest-performing category, recording 105 units sold, followed by Health with 74 units and Toys and Games with 65 units. Office 
Supplies recorded the lowest sales volume at 24 units, while Beauty and Personal Care, Sports and Outdoors, and Home and Kitchen also recorded 
relatively lower volumes. This indicates that customer demand is particularly strong for Electronics and Health-related products.

- Business Insight

PrimeNest Retail should prioritize Electronics and Health products for inventory planning and marketing, while reviewing the performance of Office 
Supplies to identify opportunities to improve demand through pricing, promotion or product selection.
```

# - Revenue by Product

```python
#Moving from quantity sold to money generated.
```

```python
"""
Revenue will be calculated as:

Quantity × Unit Price − Discount
```

```python
#This will show which individual products generate the most revenue, which can be different from the products with the highest quantity sold.
```

```python
product_revenue = (
    order_details.assign(
        revenue=lambda x: (x['quantity'] * x['unit_price']) - x['discount']
    )
    .groupby('product_id')['revenue']
    .sum()
    .reset_index()
    .merge(
        products[['product_id', 'product_name']],
        on='product_id',
        how='left'
    )
    [['product_id', 'product_name', 'revenue']]
    .sort_values('revenue', ascending=False)
)

product_revenue
```

```python
"""
- Revenue by Product: Observation
Revenue is heavily concentrated in a few products. HP Laptop 15 generated the highest revenue at 24.48 million, followed by Samsung Galaxy A55 at 12.60 
million. Office Chair ranked third at approximately 1.63 million, while Children Story Book generated the lowest revenue at approximately 134,891. This 
shows that high-value products, particularly the two leading electronics products, contribute substantially more revenue than most other products.

- Business Insight

PrimeNest Retail should prioritize the availability of HP Laptop 15 and Samsung Galaxy A55, as disruptions in the supply of these products could have a 
significant impact on revenue. The business should also investigate opportunities to improve the performance of lower-revenue products.

- And there's an interesting point here: the products selling the most units aren't necessarily generating the most revenue. For example, Wireless Mouse 
sold the most units (39), but HP Laptop 15 generated by far the highest revenue. That's an important distinction we'll use later in the project.
```

# - Monthly Order Volume & Revenue

```python
#Step 1: Create monthly order/revenue analysis

monthly_sales = (
    orders[['order_id', 'order_date']]
    .merge(
        order_details[['order_id', 'quantity', 'unit_price', 'discount']],
        on='order_id',
        how='left'
    )
    .assign(
        revenue=lambda x: (x['quantity'] * x['unit_price']) - x['discount'],
        month=lambda x: x['order_date'].dt.to_period('M').astype(str)
    )
    .groupby('month')
    .agg(
        order_volume=('order_id', 'nunique'),
        revenue=('revenue', 'sum')
    )
    .reset_index()
    .sort_values('month')
)

monthly_sales
```

```python
"""
- Monthly Order Volume & Revenue: Observation

Monthly order activity and revenue show considerable variation throughout 2025. June recorded the highest order volume with 14 orders and generated 
approximately 10.28 million in revenue. November generated the second-highest revenue at approximately 9.30 million despite having only 7 orders, 
indicating that revenue is influenced not only by order volume but also by the value of products purchased. December recorded the lowest order volume 
(3) and revenue (approximately 332,454).

- Business Insight

The results show that higher order volume does not always translate into higher revenue. PrimeNest Retail should therefore monitor both order frequency 
and order value when evaluating monthly sales performance.
```

# - Employee Sales Performance

```python
#Answers: Which employees handle the highest number of orders and generate the most revenue?
```

```python
employee_sales = (
    orders[['order_id', 'employee_id']]
    .merge(
        order_details[['order_id', 'quantity', 'unit_price', 'discount']],
        on='order_id',
        how='left'
    )
    .assign(
        revenue=lambda x: (x['quantity'] * x['unit_price']) - x['discount']
    )
    .groupby('employee_id')
    .agg(
        order_count=('order_id', 'nunique'),
        revenue=('revenue', 'sum')
    )
    .reset_index()
    .sort_values('revenue', ascending=False)
    .merge(
        employees[['employee_id', 'first_name', 'last_name']],
        on='employee_id',
        how='left'
    )
)

employee_sales
```

# - Customer Spending Analysis

```python
#Question is: Which customers generate the most revenue for PrimeNest Retail?
```

```python
#To answer the above question, orders, order_details, and customers will be combined to calculate each customer's total spending.
```

```python
customer_spending = (
    orders[['order_id', 'customer_id']]
    .merge(
        order_details[['order_id', 'quantity', 'unit_price', 'discount']],
        on='order_id',
        how='left'
    )
    .assign(
        spending=lambda x: (x['quantity'] * x['unit_price']) - x['discount']
    )
    .groupby('customer_id')
    .agg(
        order_count=('order_id', 'nunique'),
        total_spending=('spending', 'sum')
    )
    .reset_index()
    .merge(
        customers[['customer_id', 'first_name', 'last_name']],
        on='customer_id',
        how='left'
    )
    .sort_values('total_spending', ascending=False)
)

customer_spending
```

```python
"""
- Customer Spending: Observation

Customer spending varies significantly across PrimeNest Retail's customer base. Tunde Adebayo is the highest-value customer, generating approximately 
6.62 million from 9 orders, followed by Mercy Ibrahim (5.59 million) and Joy Nnamdi (5.47 million). At the lower end, Chinedu Okafor generated 
approximately 95,467 from 3 orders. The results show that a relatively small group of customers contributes substantially more revenue than others.

- Business Insight

PrimeNest Retail should prioritize customer retention and personalized engagement for high-value customers, while exploring strategies such as targeted 
promotions and product recommendations to increase spending among lower-value customers.
```

# - Inventory Analysis

```python
"""
It will answer the qestion: 

    Which products have the lowest available inventory, and how does their stock compare with their reorder levels?
```

```python
inventory_analysis = (
    inventory[['product_id', 'quantity_available']]
    .merge(
        products[['product_id', 'product_name', 'reorder_level']],
        on='product_id',
        how='left'
    )
    .assign(
        stock_buffer=lambda x: x['quantity_available'] - x['reorder_level']
    )
    .sort_values('stock_buffer')
)

inventory_analysis
```

```python
"""
- Inventory Analysis: Observation

Inventory levels are generally above their respective reorder levels, with all products maintaining a positive stock buffer. Office Chair has the lowest 
stock buffer at 20 units, followed by HP Laptop 15 and Protein Powder at 25 units each. Wireless Mouse has the highest available stock at 120 units and 
the largest stock buffer of 100 units. No product currently falls below its reorder level.

- Business Insight

Although no immediate reorder requirement exists, Office Chair should receive the closest inventory monitoring because it has the smallest stock buffer. 
High-demand products such as HP Laptop 15 and Protein Powder should also be monitored closely to prevent future stock shortages.
```
